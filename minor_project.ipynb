1.Create a menu-driven calculator that performs multiple mathematical operations using Python and
NumPy.


```python
import numpy as np
#menu
while True:
  print("1.Addition\n")
  print("2.Substration")
  print("3.Multiplication")
  print("4.Division")
  print("5.Power calcultaion")
  print("6.modulus")
  print("0.exit")

  choice = int(input("enter your choice: "))

  n = int(input("how many numbers ? : "))
  num = []

  for i in range(n):
    num.append(float(input("enter your numbers: ")))
  #addition
  if choice == 1:
    arr = np.array(num)
    sum = np.sum(arr)
    print("the sum is : ",sum)
  #substraction
  elif choice == 2:
    sub = num[0]
    for i in num[1:]:
      sub -= i
    print("substraion: ",sub)
  #multiplication
  elif choice == 3:
    mul = 1
    for i in num[1:]:
      mul *= i
    print("multiplication: ",mul)
  #division
  elif choice == 4:
    div = num[0]
    for i in num[1:]:
      if i == 0:
        print("please enter other number , zero is not allowed in division.")
        break
      else:
        div /= i
    print("division : ",div)
  #power calculation
  elif choice == 5:
    base = float(input("enter the base number : "))
    power = float(input("enter the power of base number: "))
    p_c = base**power
    print("power calculation : ",p_c)
  #modulus
  elif choice == 6:
    if num[1] == 0:
      print("modulus by zero not allowed, please enter another number : ")
      break
    else:
      print("modulus : ", num[0]%num[1])

  #exit
  elif choice == 0:
    print("thanks for using the calculator, exit!")
    break

  else:
    print("please enter valid choice.")

  cont = input("do you want to continue? yes/no: ")
  if cont == 'no':
    break












```
